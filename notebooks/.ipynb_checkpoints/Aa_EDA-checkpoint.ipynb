```python
import re
from collections import Counter

def build_dynamic_gazetteer(df, top_n=30):
    """
    Scans the dataset to automatically discover the most frequent 
    trailing geographic identifiers (cities/states) per country.
    """
    gazetteer = {}
    
    # Group by country to isolate geographic terms
    for country in df['country'].dropna().unique():
        country_str = str(country).lower()
        trailing_tokens = []
        
        subset = df[df['country'] == country]['clean_address'].dropna()
        
        for address in subset:
            # Extract all alphabetic words
            words = re.findall(r'\b[a-z]{2,}\b', address)
            if words:
                # Grab the last word (highly likely to be state/city)
                trailing_tokens.append(words[-1])
                # Optionally grab the second to last word as well
                if len(words) > 1:
                    trailing_tokens.append(words[-2])
                    
        # Find the most common trailing words for this country
        most_common = [word for word, count in Counter(trailing_tokens).most_common(top_n)]
        
        # Remove generic address words that might have slipped to the end
        stop_words = {'st', 'rd', 'ave', 'blvd', 'ltd', 'corp', 'inc', 'floor', 'building'}
        gazetteer[country_str] = set(most_common) - stop_words
        
    return gazetteer

def extract_dynamic_geo_key(address, country_str, dynamic_gazetteer):
    """
    Applies the learned gazetteer to cluster the records.
    """
    if pd.isna(address) or pd.isna(country_str):
        return f"{country_str}_unknown"
    
    address_str = str(address).lower()
    
    # Check if the address contains any of the automatically learned geographic keywords for this country
    if country_str in dynamic_gazetteer:
        for loc in dynamic_gazetteer[country_str]:
            if re.search(rf'\b{loc}\b', address_str):
                return f"{country_str}_{loc}"
                
    return f"{country_str}_unknown"

# --- Integration into your Pipeline ---
# 1. Combine all your sources temporarily to learn the vocabulary
all_data = pd.concat([s1_normalized, s2_normalized, s3_normalized])

# 2. Build the gazetteer dynamically (Takes < 2 seconds)
learned_gazetteer = build_dynamic_gazetteer(all_data, top_n=40)
print("Learned US Gazetteer:", learned_gazetteer.get('us'))

# 3. Apply it to create the geo_block_key
s1_normalized['geo_block_key'] = s1_normalized.apply(
    lambda row: extract_dynamic_geo_key(row['clean_address'], str(row['country']).lower(), learned_gazetteer), 
    axis=1
)
```
